# Compare four chart-explanation strategies

This runs the same fixed-seed CharXiv sample through direct, structured, chain-of-thought, and verification prompts. Verification drafts an answer and then checks it against the chart in a second LM Studio call. Each result records the strategy, prompts, model answer, reasoning output, reference answer, and saved chart path. A progress bar tracks LM Studio requests.

Install the repository's `requirements.txt`, load a vision model in LM Studio, and start its local server before running. The output is saved under the repository's ignored `outputs/` directory.

```python
import sys
from pathlib import Path

WORKING_DIR = Path.cwd().resolve()
PROJECT_ROOT = next(
    (path for path in (WORKING_DIR, *WORKING_DIR.parents)
     if (path / 'Experiments' / 'charxiv_prompt_strategies.py').is_file()),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError('Could not find Experiments/charxiv_prompt_strategies.py from the notebook working directory.')
sys.path.insert(0, str(PROJECT_ROOT / 'Experiments'))
from charxiv_prompt_strategies import run_pipeline, STRATEGY_PROMPTS

N = 5
SEED = 42
MODEL = 'google/gemma-4-e4b'  # Use the identifier shown by your LM Studio server.
API_URL = 'http://localhost:1234/api/v1/chat'
OUTPUT = PROJECT_ROOT / 'outputs' / 'prompt_strategies' / 'charxiv_answers_v2.json'
RESUME = False  # Set True after an interrupted run to keep completed results.
```

```python
output_path = run_pipeline(
    output_path=OUTPUT, n=N, seed=SEED, model=MODEL, api_url=API_URL, resume=RESUME
)
print(f'Saved {N * len(STRATEGY_PROMPTS)} responses to {output_path}')
```

```python
import json

run = json.loads(output_path.read_text(encoding='utf-8'))
print('Status:', run['status'], '| Results:', len(run['results']))
run['results'][0]
```
